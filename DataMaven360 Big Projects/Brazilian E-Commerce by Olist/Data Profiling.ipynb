```python
import pandas as pd
from pathlib import Path
```

```python
df = pd.read_csv (Path.cwd() / "Brazilian E-Commerce Public Dataset by Olist" / "olist_customers_dataset.csv")
df.head()
```

```python
df.info()
```

```python
summary = []
for col in df.columns:
    total_rows = len(df)
    null_count = df[col].isnull().sum()
    col_dict = {
        "Column": col,
        "Data Type": df[col].dtype,
        "Total Rows": total_rows,
        "Non-Null Count": df[col].count(),
        "Null Count": null_count,
        "Null %": null_count / total_rows * 100,
        "Distinct Values": df[col].nunique()
    }
    summary.append(col_dict)    

# Convert the list of dictionaries into a DataFrame — once, at the end
summary_df = pd.DataFrame(summary)
print(summary_df)
```

```python
#print(df.duplicated(subset="customer_unique_id").sum())
f_row_count = {
    "rows": len(df),
    "columns": len(df.columns)
}
```

```python
f_row_count = {
    "rows": len(df),
    "columns": len(df.columns)
}

df_row_count = f_row_count
df_column_summary = summary_df

rows= df_row_count["rows"]
columns= df_row_count["columns"]
duplicate_Rows= df.duplicated().sum()
columns_With_Nulls= len(df_column_summary.query("`Null Count` > 0"))
completely_Empty_Columns= len(df_column_summary.query("`Null %` == 100"))
potential_Primary_Keys= (df_column_summary["Distinct Values"] == df_column_summary["Total Rows"]).sum()
potential_Date_Columns= len(df_column_summary.query("`Data Type` == 'object'"))
potential_Numeric_Columns= len(df_column_summary.query("`Data Type` == 'int64'"))
potential_Categorical_Columns= len(df_column_summary.query("`Data Type` == 'string'"))
Health_Score = None

health_summary = { 
    "Rows": rows, 
    "Columns": columns, 
    "Duplicate Rows": duplicate_Rows, 
    "Columns With Nulls": columns_With_Nulls, 
    "Completely Empty Columns": completely_Empty_Columns, 
    "Potential Primary Keys": potential_Primary_Keys, 
    "Potential Date Columns": potential_Date_Columns, 
    "Potential Numeric Columns": potential_Numeric_Columns, 
    "Potential Categorical Columns": potential_Categorical_Columns, 
    "Health Score": Health_Score }
print(health_summary)
```

```python
zip_counts = (
    df.groupby("customer_unique_id")["customer_zip_code_prefix"]
      .nunique()
      .reset_index(name="distinct_zip_count")
).sort_values("distinct_zip_count",ascending=False)
print(zip_counts)
```

```python
zip_counts = (
    df.groupby("customer_unique_id")["customer_city"]
      .nunique()
      .reset_index(name="distinct_city_count")
).sort_values("distinct_city_count",ascending=False)
print(zip_counts)
```

```python
zip_counts = (
    df.groupby("customer_unique_id")["customer_state"]
      .nunique()
      .reset_index(name="distinct_state_count")
).sort_values("distinct_state_count",ascending=False)

print(len(zip_counts[zip_counts["distinct_state_count"] > 1]))
```
